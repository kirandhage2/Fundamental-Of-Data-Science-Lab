```python
import numpy as np
print("NumPy version:", np.__version__)
```

```python

arr1 = np.array([1, 2, 3, 4, 5])
print("1D array:", arr1)

arr2 = np.array([[1, 2, 3], [4, 5, 6]])
print("2D array:\n", arr2)


arr3 = np.array([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])
print("3D array:\n", arr3)
```

```python
# arange -> similar to Python range() but returns an array
range_arr = np.arange(0, 10, 2)  
print("arange array:", range_arr)

lin_arr = np.linspace(0, 1, 5)    
print("linspace array:", lin_arr)

rand_arr = np.random.rand(2, 3)     
randint_arr = np.random.randint(1, 100, size=(5, 3))  
print("Random float array:\n", rand_arr)

print("Random integer array:\n", randint_arr)
```

```python
exercise1_a = np.zeros((4, 4))
exercise1_b = np.arange(10, 51, 5)
print(exercise1_a)
print(exercise1_b)
```

```python
sample = np.array([[1, 2, 3], [7,8, 9]])

print("Array:\n", sample)
print("Shape:", sample.shape)      
print("Dimensions:", sample.ndim)  
print("Size:", sample.size)       
print("Data type:", sample.dtype)
```

```python
x = np.array([13, 20, 35, 40])
y = np.array([1, 2, 3, 4])

print("Addition:", x + y)
print("Subtraction:", x - y)
print("Multiplication:", x * y)
print("Division:", x / y)
print("Power:", x ** 2)
```

```python
data = np.array([4, 8, 15, 16, 25, 40])

print("Sum:", np.sum(data))
print("Mean:", np.mean(data))
print("Minimum:", np.min(data))
print("Maximum:", np.max(data))
print("Standard deviation:", np.std(data))
print("Variance:", np.var(data))
```

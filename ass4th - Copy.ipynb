```python
import pandas as pd
import numpy as np

df = pd.read_csv('CustomerWaitTimes.csv')

print(df.head())
print(df.describe())

range_a = df['Branch_A'].max() - df['Branch_A'].min()
range_b = df['Branch_B'].max() - df['Branch_B'].min()

print('Range - Branch A:', range_a)
print('Range - Branch B:', range_b)

var_a = df['Branch_A'].var()
var_b = df['Branch_B'].var()

print('Variance - Branch A:', var_a)
print('Variance - Branch B:', var_b)

std_a = df['Branch_A'].std()
std_b = df['Branch_B'].std()

print('Std Dev - Branch A:', std_a)
print('Std Dev - Branch B:', std_b)

cv_a = (std_a / df['Branch_A'].mean()) * 100
cv_b = (std_b / df['Branch_B'].mean()) * 100

print('CV% - Branch A:', cv_a)
print('CV% - Branch B:', cv_b)

if std_a < std_b:
    print('Branch A has more consistent waiting times than Branch B')
else:
    print('Branch B has more consistent waiting times than Branch A')
```
